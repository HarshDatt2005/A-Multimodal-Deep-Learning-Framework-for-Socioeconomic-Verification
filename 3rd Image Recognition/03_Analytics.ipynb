```python
import pandas as pd

# Using read_excel for the .xlsx file
file_path = 'ML_Training_Dataset_Cleaned.xlsx'
df = pd.read_excel(file_path)

# Columns to analyze for imbalance
columns_to_analyze = [
    'Image_Context', 'Housing_Category', 'Ext_Roof_Type', 'Ext_Wall_Type', 
    'Int_Floor_Material', 'Int_Wall_Finish', 'Overall_Structural_Score', 'Ext_Structural_Condition'
]

# Calculating and printing the imbalance for each column
for col in columns_to_analyze:
    if col in df.columns:
        counts = df[col].value_counts(dropna=False)
        percentages = df[col].value_counts(normalize=True, dropna=False) * 100
        
        imbalance_report = pd.DataFrame({
            'Total Count': counts,
            'Percentage (%)': percentages.round(2)
        })
        print(f"--- Class Distribution for '{col}' ---")
        print(imbalance_report)
        print("-" * 50 + "\n")
    else:
        print(f"Column '{col}' not found in the dataset.\n")
```
