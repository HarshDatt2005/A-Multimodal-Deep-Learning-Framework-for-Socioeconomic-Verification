```python
import os
import pandas as pd

image_folder = "./IMAGES/"
excel_path = "ML_Training_Dataset.xlsx"

# Getting the list of exact files currently in our folder
valid_extensions = ('.jpg', '.jpeg', '.png')
actual_files = set([f for f in os.listdir(image_folder) if f.lower().endswith(valid_extensions)])

# Loading our current Excel dataset
df = pd.read_excel(excel_path)

# Finding Duplicates
duplicates = df[df.duplicated(subset=['Image_Name'], keep=False)]
print(f"Found {len(duplicates)} duplicate entries in the Excel file.")

# Finding Ghost Images (In Excel, but missing from the folder)
excel_images = set(df['Image_Name'])
ghost_images = excel_images - actual_files
if ghost_images:
    print(f"Found {len(ghost_images)} 'Ghost' records (in Excel but missing from folder):")
    for ghost in list(ghost_images)[:5]: 
        print(f"   - {ghost}")

# --- THE CLEANUP ---
print("\n--- Synchronizing Dataset ---")

# Step A: Dropping exact duplicates, keeping the most recent one
df_cleaned = df.drop_duplicates(subset=['Image_Name'], keep='last')

# Step B.1: Removing any rows where the image no longer exists in the folder
df_cleaned = df_cleaned[df_cleaned['Image_Name'].isin(actual_files)]

# Step C: Saving the perfectly synced file
cleaned_excel_path = "ML_Training_Dataset_Cleaned.xlsx"
df_cleaned.to_excel(cleaned_excel_path, index=False)

print(f"Done! Cleaned dataset has {len(df_cleaned)} records.")
print(f"Matches folder count ({len(actual_files)})? {len(df_cleaned) == len(actual_files)}")
print(f"Saved safely to: {cleaned_excel_path}")
```
