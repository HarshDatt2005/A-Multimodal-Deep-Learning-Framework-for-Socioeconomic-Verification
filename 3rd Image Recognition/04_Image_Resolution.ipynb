```python
import os
from PIL import Image

def get_average_image_size(folder_path):
    print(f"Scanning directory: {folder_path}...")
    valid_extensions = ('.jpg', '.jpeg', '.png')
    
    # Getting all image files
    files = [f for f in os.listdir(folder_path) if f.lower().endswith(valid_extensions)]
    
    if not files:
        print("No images found in the specified directory.")
        return
        
    total_width = 0
    total_height = 0
    min_width = float('inf')
    min_height = float('inf')
    max_width = 0
    max_height = 0
    
    for index, filename in enumerate(files):
        filepath = os.path.join(folder_path, filename)
        
        try:
            # We use 'with' so the file closes properly to save RAM
            with Image.open(filepath) as img:
                w, h = img.size
                
                total_width += w
                total_height += h
                
                # Track the extremes
                if w < min_width: min_width = w
                if h < min_height: min_height = h
                if w > max_width: max_width = w
                if h > max_height: max_height = h
                
        except Exception as e:
            print(f"Error reading {filename}: {e}")

    avg_width = total_width / len(files)
    avg_height = total_height / len(files)
    
    print("\n--- 📊 RESOLUTION REPORT ---")
    print(f"Total Images Scanned : {len(files)}")
    print(f"Average Resolution   : {int(avg_width)} x {int(avg_height)} pixels")
    print(f"Smallest Image       : {min_width} x {min_height} pixels")
    print(f"Largest Image        : {max_width} x {max_height} pixels")
    print("----------------------------\n")

# RUNNING THE SCANNER ON OUR RAW IMAGES
get_average_image_size("./IMAGES/")
```
