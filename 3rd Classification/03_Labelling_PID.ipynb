```python
import pandas as pd

# Defining the file names
source_pid_file = "processed_district_data_with_property_ids.xlsx"
target_labeled_file = r"C:\Users\Harsh Datt\Data Science\2nd Classification\labeled_final_dataset.xlsx"
output_file = "labeled_final_dataset_with_pids.xlsx"

print("Loading datasets...")
# Loading the datasets
df_pids = pd.read_excel(source_pid_file)
df_labeled = pd.read_excel(target_labeled_file)

# Creating a dictionary mapping Family IDs to Property IDs
# We drop duplicates just in case there are multiple rows for the same family in the source file
mapping_df = df_pids[['hasfamilyid', 'Property_ID']].drop_duplicates()
family_to_pid_map = dict(zip(mapping_df['hasfamilyid'], mapping_df['Property_ID']))

# Applying the mapping to the labeled dataset
# Inserting it at position 0 (the very first column)
df_labeled.insert(0, 'Property_ID', df_labeled['hasfamilyid'].map(family_to_pid_map))

# Optional: Checking how many successfully matched
matched_count = df_labeled['Property_ID'].notna().sum()
total_count = len(df_labeled)
print(f"Matched {matched_count} out of {total_count} rows.")

# Saving the updated dataset
print(f"Saving updated file to {output_file}...")
df_labeled.to_excel(output_file, index=False)
print("Process completed successfully!")
```
