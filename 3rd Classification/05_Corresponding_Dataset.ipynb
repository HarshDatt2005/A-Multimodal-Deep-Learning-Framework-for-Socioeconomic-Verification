```python
import pandas as pd

# Configuration
input_file = "families_mapped_to_images.xlsx"
output_file = "families_mapped_to_images_filtered_for_analysis.xlsx"

print(f"Loading {input_file}...")
df = pd.read_excel(input_file)

# Defining the exact columns to keep
columns_to_keep = [
    "Property_ID", 
    "hasfamilyid", 
    "r_u", 
    "BPL_Target", 
    "Image_Name", 
    "Image_Context", 
    "Housing_Category", 
    "Ext_Stories", 
    "Ext_Roof_Type", 
    "Ext_Wall_Type", 
    "Int_Floor_Material", 
    "Int_Wall_Finish",
    "Visible_Assets_For_Income", 
    "Overall_Structural_Score", 
    "Ext_Structural_Condition"
]

# Checking for exact matches in the dataset
actual_columns = df.columns.tolist()
matched_columns = [col for col in columns_to_keep if col in actual_columns]
missing_columns = [col for col in columns_to_keep if col not in actual_columns]

if missing_columns:
    print(f"\nWARNING: The following columns were not found in the file:")
    for col in missing_columns:
        print(f" - {col}")
    print("\nPlease double-check the exact spelling of these columns in your Excel file.")

# Filtering the DataFrame to keep only the matched columns
df_filtered = df[matched_columns]

# Saving the output
print(f"\nSaving filtered dataset with {len(matched_columns)} columns...")
df_filtered.to_excel(output_file, index=False)
print(f"Successfully saved to {output_file}!")
```
