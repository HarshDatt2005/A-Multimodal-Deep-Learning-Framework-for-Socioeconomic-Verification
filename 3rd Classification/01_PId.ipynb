```python
import pandas as pd
import random
import string

# 1. Reading the Excel file
file_path = r"C:\Users\Harsh Datt\Data Science\2nd Classification\100-district-wise-data-160626.xlsx"
df = pd.read_excel(file_path)

# 2. Extracting all unique Family IDs
unique_family_ids = df['hasfamilyid'].unique()

# 3. Creating a function to generate unique Property IDs
def generate_unique_property_ids(num_ids):
    generated = set()
    while len(generated) < num_ids:
        # Pattern: 1 digit + 3 uppercase letters + 4 digits
        first_digit = str(random.randint(1, 9))
        letters = ''.join(random.choices(string.ascii_uppercase, k=3))
        last_digits = f"{random.randint(0, 9999):04}"
        
        full_id = first_digit + letters + last_digits
        generated.add(full_id)
        
    return list(generated)

# 4. Generating exactly enough unique Property IDs for all families
unique_property_ids = generate_unique_property_ids(len(unique_family_ids))

# 5. Creating a mapping (dictionary) matching each Family ID to a Property ID
family_to_property_map = dict(zip(unique_family_ids, unique_property_ids))

# 6. Applying the mapping to the DataFrame, inserting it as the first column
df.insert(0, 'Property_ID', df['hasfamilyid'].map(family_to_property_map))

# 7. Saving the results to a new file
output_path = "district_data_with_property_ids.xlsx"
df.to_excel(output_path, index=False)
```
