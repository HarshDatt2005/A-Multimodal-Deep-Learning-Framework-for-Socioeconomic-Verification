```python
import pandas as pd

def create_family_level_dataset(input_excel_path, output_excel_path):
    print(f"Loading data from {input_excel_path}...")
    
    # Reading the original Excel file
    raw_df = pd.read_excel(input_excel_path)
    
    # Cleaning column names to prevent errors from accidental spaces
    raw_df.columns = raw_df.columns.str.strip()
    
    # Collapsing to Family Level (One row per FID)
    print("Aggregating individual members into family units...")
    family_df = raw_df.groupby('FID').agg(
        Family_Size=('Name', 'count'),       # Count how many members share the FID
        Primary_Contact=('Mob', 'first'),    # Grab the first mobile number listed
        Address=('Address', 'first')         # Grab the first address listed
    ).reset_index()
    
    # Exporting to a new Excel file
    print(f"Saving aggregated dataset to {output_excel_path}...")
    family_df.to_excel(output_excel_path, index=False, engine='openpyxl')
    
    print("\n--- Process Complete! ---")
    print(f"Total Unique Families: {len(family_df)}")
    print(family_df.head())

# Execution
if __name__ == "__main__":
    # Define your input and output file names
    INPUT_FILE = "sample BPL  sector 10 PKL.xlsx" 
    OUTPUT_FILE = "Aggregated_BPL_Families.xlsx"
    
    # Run the function
    create_family_level_dataset(INPUT_FILE, OUTPUT_FILE)
```
