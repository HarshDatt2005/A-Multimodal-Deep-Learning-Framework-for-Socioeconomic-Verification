```python
import pandas as pd

# LOADING THE MANUALLY VALIDATED DATASETS
print("Loading datasets...")
df_final = pd.read_excel(r'C:\Users\Harsh Datt\Data Science\2nd Classification\labeled_final_dataset.xlsx')
df_master = pd.read_excel(r'C:\Users\Harsh Datt\Data Science\3rd Classification\labeled_master_dataset.xlsx')

# MERGING ON UNIQUE IDENTIFIER
# We merge on 'hasfamilyid' to ensure we are comparing the exact same families.
# The suffixes parameter renames the overlapping BPL_Target columns.
merged_df = pd.merge(
    df_final[['hasfamilyid', 'BPL_Target']], 
    df_master[['hasfamilyid', 'BPL_Target']], 
    on='hasfamilyid', 
    suffixes=('_final', '_master')
)

# CALCULATING THE COUNTS
# Condition 1: Final = 0, Master = 0
count_0_0 = len(merged_df[(merged_df['BPL_Target_final'] == 0) & (merged_df['BPL_Target_master'] == 0)])

# Condition 2: Final = 0, Master = 1
count_0_1 = len(merged_df[(merged_df['BPL_Target_final'] == 0) & (merged_df['BPL_Target_master'] == 1)])

# Condition 3: Final = 1, Master = 0
count_1_0 = len(merged_df[(merged_df['BPL_Target_final'] == 1) & (merged_df['BPL_Target_master'] == 0)])

# Condition 4: Final = 1, Master = 1
count_1_1 = len(merged_df[(merged_df['BPL_Target_final'] == 1) & (merged_df['BPL_Target_master'] == 1)])

# PRINTING THE RESULTS
print("\n--- BPL Target Comparison Results ---")
print(f"Total matching families found: {len(merged_df)}\n")

print(f"1. Final (0) and Master (0): {count_0_0} families")
print(f"2. Final (0) and Master (1): {count_0_1} families")
print(f"3. Final (1) and Master (0): {count_1_0} families")
print(f"4. Final (1) and Master (1): {count_1_1} families")

# A quick pandas trick to view this as a matrix!
print("\n--- Summary Matrix (Cross-Tabulation) ---")
print(pd.crosstab(merged_df['BPL_Target_final'], merged_df['BPL_Target_master'], 
                  rownames=['Final_Dataset'], colnames=['Master_Dataset']))
```
